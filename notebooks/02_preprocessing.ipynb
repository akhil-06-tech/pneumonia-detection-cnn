```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import os

train_dir = os.path.join("..", "data", "chest_xray", "train")
IMG_SIZE = (150, 150)
BATCH_SIZE = 32

train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=10,
    zoom_range=0.1,
    horizontal_flip=False
)

train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='binary'
)

# Fetch a single batch to verify shapes
images, labels = next(train_generator)
print("Batch Image Tensor Shape:", images.shape)  # Should output (32, 150, 150, 3)
print("Batch Labels Shape:", labels.shape)          # Should output (32,)
print("Min pixel value:", images.min(), "| Max pixel value:", images.max())
```
