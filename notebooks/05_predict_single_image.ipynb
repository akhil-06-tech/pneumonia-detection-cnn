```python
import tensorflow as tf
from tensorflow.keras.preprocessing import image
import numpy as np
import os

# 1. Load your trained model
model_path = os.path.join("..", "models", "pneumonia_cnn_model.keras")
model = tf.keras.models.load_model(model_path)

# 2. Specify the path to the single X-ray image you want to check
# (Replace this with the actual file path of an image on your computer)
img_path = "../data/chest_xray/test/NORMAL/IM-0011-0001-0001.jpeg"  


# 3. Load and resize the image to match your model's expected input (150x150)
img = image.load_img(img_path, target_size=(150, 150))


# 4. Convert the image to a numpy array and preprocess it (rescale pixel values)
img_array = image.img_to_array(img) / 255.0
img_array = np.expand_dims(img_array, axis=0)  # Add batch dimension: (1, 150, 150, 3)


# 5. Make the prediction
prediction = model.predict(img_array)[0][0]
print(f"Raw Model Prediction Score: {prediction:.4f}")

# 6. Improved Decision Logic with a Buffer Zone
if prediction > 0.65:
    confidence = prediction * 100
    print(f"Result: PNEUMONIA Detected (Confidence: {confidence:.2f}%)")
elif prediction < 0.35:
    confidence = (1 - prediction) * 100
    print(f"Result: NORMAL (Confidence: {confidence:.2f}%)")
else:
    print(f"Result: UNCERTAIN / BORDERLINE (Score: {prediction:.4f}) - Needs doctor review.")
```
