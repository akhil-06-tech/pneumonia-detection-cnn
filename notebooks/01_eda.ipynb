```python
import os
import matplotlib.pyplot as plt
import seaborn as plt_sns
from PIL import Image

train_dir = os.path.join("..", "data", "chest_xray", "train")
val_dir = os.path.join("..", "data", "chest_xray", "val")
test_dir = os.path.join("..", "data", "chest_xray", "test")

# Count images in each category
for split_name, split_path in [("Train", train_dir), ("Validation", val_dir), ("Test", test_dir)]:
    normal_count = len(os.listdir(os.path.join(split_path, "NORMAL")))
    pneumonia_count = len(os.listdir(os.path.join(split_path, "PNEUMONIA")))
    print(f"{split_name} Set -> Normal: {normal_count} | Pneumonia: {pneumonia_count} | Total: {normal_count + pneumonia_count}")
```

```python
def show_sample_images(folder_path, label_name):
    normal_folder = os.path.join(folder_path, "NORMAL")
    pneumonia_folder = os.path.join(folder_path, "PNEUMONIA")
    
    normal_imgs = os.listdir(normal_folder)[:2]
    pneumonia_imgs = os.listdir(pneumonia_folder)[:2]
    
    plt.figure(figsize=(10, 10))
    
    # Plot Normal samples
    for i, img_name in enumerate(normal_imgs):
        img_path = os.path.join(normal_folder, img_name)
        image = Image.open(img_path)
        plt.subplot(2, 2, i + 1)
        plt.imshow(image, cmap='gray')
        plt.title(f"NORMAL ({label_name})")
        plt.axis('off')
        
    # Plot Pneumonia samples
    for i, img_name in enumerate(pneumonia_imgs):
        img_path = os.path.join(pneumonia_folder, img_name)
        image = Image.open(img_path)
        plt.subplot(2, 2, i + 3)
        plt.imshow(image, cmap='gray')
        plt.title(f"PNEUMONIA ({label_name})")
        plt.axis('off')
        
    plt.tight_layout()
    plt.show()

show_sample_images(train_dir, "Train")
```
